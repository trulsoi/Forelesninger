## `zip` og `enumerate`

- Vi har laget eksamen og vi har stort sett vært på samme side ang. pensum
- Et lite unntak er at de andre har sett på `zip` og `enumerate` i tillegg
- Disse er praktiske innebygde funksjoner vi kan bruke sammen med for-løkker, og vi gir et

### `zip`

- `zip` lar oss "zippe" sammen flere lister slik at vi kan gå gjennom alle listene parallelt 

- `zip` tar 2 eller flere lister, og setter de sammen i grupper som  en glidlås.
- indeks 0 elementene er i gruppe med alle indeks 0 elementer, 1 med 1 osv
- Dersom listene ikke er like lange, blir får vi like mange grupper som elementer i den korteste listen

```python
a = [1,2,3,4,5]
b = ["a", "b", "c"]

zip(a,b)
```

```python
list(zip(a,b))
```

```python
for ai, bi in zip(a,b):
    print(ai,bi)
```

```python
import numpy as np

def Ek(m, v):
    return 0.5*m*v**2

def v(v0, a, t):
    return v0+a*t

t = np.linspace(0,10, 21)
vels = v(25,-9.81, t)
energi = Ek(10, vels)

for ti, vi, ei in zip(t,vels,energi):
    print(f"Ved tiden {ti} s har vi farten {vi:.2f} m/s og kinetisk energi {ei:.2f} J")
```

### `enumerate`
- `enumerate` lar oss gå over indeks og element i liste samtidig.
- Vi kan alltids bare går over elementene med `for i in range(len(«liste»))`


```python
listeri = ["a", "b", 1, "hei"]
enumerate(listeri)
```

```python
list(enumerate(listeri))
```

```python
for indeks, element in enumerate(listeri):
    print(f"Ved indeks {indeks} finner vi element {element}")
```

- Si vi vil finne ut ved hvilket tidspunkt vi har fart nærmest 0 m/s i eksempelet over

```python
vmin = np.inf
min_indeks = None

for indeks, fart in enumerate(vels):
    if np.abs(fart) < vmin:
        vmin = fart
        min_indeks = indeks

print(f"Vi har minst fart etter {t[min_indeks]} sekunder")
```
